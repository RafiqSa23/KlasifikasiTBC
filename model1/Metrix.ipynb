```python
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
# ============================================================
# KONFIGURASI PATH
# ============================================================
BASE_DIR = '../data_split'
TEST_DIR = os.path.join(BASE_DIR, 'test')
MODEL_PATH = 'mobilenetv2_tb_final.h5'   # model terbaik dari hasil training

# Ukuran input dan batch size (harus sama dengan saat training)
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
```

```python
# ============================================================
# LOAD MODEL
# ============================================================
print("Memuat model terbaik...")
model = tf.keras.models.load_model(MODEL_PATH)
print("Model berhasil dimuat.\n")
```

```python
# ============================================================
# DATA GENERATOR UNTUK TEST SET
# ============================================================
test_datagen = ImageDataGenerator(rescale=1./255)

test_generator = test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False   # penting agar urutan label tetap sesuai
)

print(f"Jumlah kelas: {test_generator.num_classes}")
print(f"Indeks kelas: {test_generator.class_indices}\n")
```

```python
# ============================================================
# EVALUASI MODEL (loss & accuracy)
# ============================================================
print("Evaluasi model pada test set...")
test_loss, test_acc = model.evaluate(test_generator, verbose=1)
print(f"Test Loss     : {test_loss:.4f}")
print(f"Test Accuracy : {test_acc:.4f}\n")
```

```python
# ============================================================
# PREDIKSI UNTUK METRIK LAINNYA
# ============================================================
print("Melakukan prediksi...")
y_pred_proba = model.predict(test_generator, verbose=1)
y_pred = np.argmax(y_pred_proba, axis=1)
y_true = test_generator.classes

# ============================================================
# METRIK LENGKAP
# ============================================================
test_prec = precision_score(y_true, y_pred)
test_rec  = recall_score(y_true, y_pred)
test_f1   = f1_score(y_true, y_pred)

print("=" * 60)
print("📊 HASIL TEST SET")
print("=" * 60)
print(f"Accuracy  : {test_acc:.4f}")
print(f"Precision : {test_prec:.4f}")
print(f"Recall    : {test_rec:.4f}")
print(f"F1-Score  : {test_f1:.4f}")
print("\n")
```
